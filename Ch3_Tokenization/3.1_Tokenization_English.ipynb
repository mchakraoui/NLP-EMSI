## 3. What is tokenization?

**Tokenization** is the process of splitting text into units called *tokens*. A token may be a word, punctuation mark, number, or another unit recognized by the tokenizer.

With spaCy, the `nlp` object processes a string and returns a `Doc` object, which contains tokens and, depending on the pipeline, linguistic annotations.

```python
text = "Tesla is looking at buying a U.S. startup for $69 million."

# Process the text with the spaCy pipeline
doc = nlp(text)

# Display the type of the returned object
print(type(doc))

# Display each token
for token in doc:
    print(token.text)
```

## 4. Inspect token properties

Each `Token` object exposes several useful properties:
- `token.text`: the exact text of the token;
- `token.shape_`: a simplified representation of its shape;
- `token.is_alpha`: indicates whether the token consists only of alphabetic characters;
- `token.is_stop`: indicates whether the token belongs to the model's stop-word list.

A stop word can still be important to the meaning of a sentence, so it should not be removed automatically in every project.

```python
for token in doc:
    print(f"Token : {token.text!r}")
    print(f"  Forme (shape_) : {token.shape_}")
    print(f"  Alphabétique ? : {token.is_alpha}")
    print(f"  Stopword ?     : {token.is_stop}")
    print("-" * 35)
```

## 5. Access a token by its index

Tokens in a `Doc` are indexed from `0`, just like elements in a Python list. `doc[0]` refers to the first token.

```python
# Display the first tokens with their indices
for i, token in enumerate(doc[:8]):
    print(i, "->", token.text)
```

## 6. Select a group of tokens with `Span`

A slice such as `doc[2:5]` returns a `Span` object, which is a contiguous sequence of tokens. The end index is excluded: `doc[2:5]` contains the tokens at indices 2, 3, and 4.

```python
# Select the tokens at indices 2, 3, and 4
span = doc[2:5]

print("Span text:", span.text)
print("Type :", type(span))
```

## 7. Tokenization of punctuation, contractions, and URLs

spaCy's tokenizer applies language-specific rules. For example, it may separate some punctuation marks while keeping units such as email addresses together, depending on its rules.

Inspect the output rather than assuming how a string will be split.

```python
examples = [
    '"We\'re moving to L.A.!"',
    "We're here to help! Send snail-mail, email support@oursite.com or visit http://www.oursite.com!",
    "A 5km NYC cab ride costs $10.30",
    "Let's visit St. Louis in the U.S. next year."
]

for example in examples:
    print("Text:", example)
    print("Tokens:", [token.text for token in nlp(example)])
    print("-" * 60)
```

## 8. Tokenization does not automatically modify the original text

The source text remains a string. The `Doc` is a tokenized representation of that text.

**Important:** Tokens in a `Doc` object cannot be replaced using an assignment such as `doc[3] = ...`. If you want to modify the text, edit the string first and then process it again.

```python
text1 = "My dinner was horrible."
text2 = "Your dinner was delicious."

doc1 = nlp(text1)
doc2 = nlp(text2)

print("Text 1:", [token.text for token in doc1])
print("Text 2:", [token.text for token in doc2])

# To modify the text, edit the string and run the pipeline again
revised_text = text1.replace("horrible", "delicious")
revised_doc = nlp(revised_text)

print("Modified text:", revised_text)
print("New tokens:", [token.text for token in revised_doc])
```

## 9. Tokenization with NLTK

NLTK also provides tokenizers. Depending on the version, `word_tokenize()` may require the `punkt` and `punkt_tab` resources.

NLTK's output may differ from spaCy's because the two libraries use different rules.

```python
import nltk

# Download tokenization resources required by recent versions
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)

from nltk.tokenize import word_tokenize

example_text = """
Hello Mr. Smith, how are you doing today? The weather is great,
and Python is awesome. The sky is pinkish-blue. You shouldn't eat cardboard.
"""

tokens_nltk = word_tokenize(example_text, language="english")
print(tokens_nltk)
```

## 10. Compare `str.split()`, NLTK, and spaCy

`str.split()` mainly splits text at whitespace. It does not handle punctuation specially. NLP libraries apply more sophisticated tokenization rules.

There is no single universally correct tokenization for every task: choose the method that fits your application.

```python
text = "Hello Mr. Smith! You shouldn't pay $10.30."

# Simple method: split on whitespace
print("str.split():", text.split())

# Tokenisation NLTK
print("NLTK       :", word_tokenize(text, language="english"))

# Tokenisation spaCy
print("spaCy      :", [token.text for token in nlp(text)])
```

## 11. Tokenization of a longer text

For longer texts, tokenization lets you iterate through a document, count tokens, or compare segments. The following example uses a literary excerpt for demonstration.

```python
example_text = """
Thomas Gradgrind, sir. A man of realities. A man of facts and calculations.
A man who proceeds upon the principle that two and two are four.
"""

doc_long = nlp(example_text)

print("Number of tokens:", len(doc_long))
print("First 20 tokens:", [token.text for token in doc_long[:20]])

# Découper le texte en phrases à l'aide du pipeline spaCy
for sentence in doc_long.sents:
    print("-", sentence.text)
```

## 12. Tokenization of Arabic text

The `en_core_web_sm` model is an **English** model. It can produce tokens from Arabic characters, but its linguistic annotations are not suitable for Arabic.

To demonstrate tokenization only, we can create an empty spaCy pipeline with the default tokenizer. For Arabic linguistic analysis, choose a model or tool designed for that language.

```python
arabic_text = "يعد الذكاء الاصطناعي من العلوم التي يتسارع التطور فيها منذ عام 2005."

# Empty spaCy pipeline: useful for demonstrating tokenization without an English model
nlp_blank_ar = spacy.blank("ar")
doc_ar = nlp_blank_ar(arabic_text)

for token in doc_ar:
    print(
        f"Token: {token.text!r} | "
        f"Shape: {token.shape_} | "
        f"Alphabetic: {token.is_alpha}"
    )
```

## 13. Tokenize Arabic text with NLTK

NLTK can also split Arabic text. The result depends on the tokenizer used; this operation does not perform Arabic morphological or syntactic analysis.

```python
arabic_text = """
أبو عبد الله محمد بن موسى الخوارزمي عالم رياضيات وفلك وجغرافيا مسلم.
يعتبر من أوائل علماء الرياضيات المسلمين.
"""

# For this Arabic example, use a simple whitespace-based tokenizer.
# It does not perform morphological analysis and keeps punctuation attached to words.
from nltk.tokenize import WhitespaceTokenizer

tokens_ar = WhitespaceTokenizer().tokenize(arabic_text)
print(tokens_ar)
```

## 14. Summary of objects and properties

| Element | Meaning |
|---|---|
| `nlp(text)` | Applies the spaCy pipeline and returns a `Doc` |
| `Doc` | Tokenized representation of the text |
| `Token` | An individual unit in the document |
| `Span` | A contiguous sequence of tokens |
| `token.text` | Exact text of the token |
| `token.shape_` | Simplified shape of the token |
| `token.is_alpha` | True if the token is alphabetic |
| `token.is_stop` | True if the token is a stop word in the model |
| `word_tokenize(text)` | Tokenization with NLTK |
| `text.split()` | Simple splitting on whitespace |
