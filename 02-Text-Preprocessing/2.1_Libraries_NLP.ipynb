### 2.3 Tokenisation : découper un texte

La **tokenisation** consiste à découper un texte en unités appelées *tokens*. Selon le besoin, les tokens peuvent être des phrases ou des mots.

Exemple : le texte ci-dessous contient deux phrases.


```python
from nltk.tokenize import sent_tokenize, word_tokenize

text = "Natural Language Processing is fascinating. It helps computers work with human language."

# Découper le texte en phrases
sentences = sent_tokenize(text, language="english")
print("Phrases :", sentences)

# Découper le texte en mots et signes de ponctuation
tokens = word_tokenize(text, language="english")
print("Tokens :", tokens)
```

**À observer :** la tokenisation en mots peut conserver la ponctuation comme tokens distincts. Le résultat dépend de la méthode et de la langue utilisée.

### 2.4 Stopwords — mots vides

Les *stopwords* sont des mots très fréquents (par exemple *the*, *is*, *and*) que l’on retire parfois pour certaines tâches. Il ne faut toutefois pas les supprimer automatiquement : ils peuvent être importants pour l’analyse du sens, notamment dans la négation (*not good*).


```python
from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))

# Garder les tokens qui contiennent au moins un caractère alphabétique
# et qui ne sont pas des stopwords
filtered_tokens = [
    token for token in tokens
    if token.isalpha() and token.lower() not in english_stopwords
]

print("Tokens d'origine :", tokens)
print("Après filtrage   :", filtered_tokens)
```

## 3. spaCy — analyse linguistique

spaCy est une bibliothèque conçue pour construire des pipelines NLP efficaces. Elle permet notamment la tokenisation, l’analyse morphosyntaxique, la reconnaissance d’entités nommées et l’analyse des dépendances.

### 3.1 Charger un modèle de langue

Installer le paquet `spacy` ne suffit pas : il faut aussi installer un **modèle linguistique**. Le modèle `en_core_web_sm` est un petit modèle anglais adapté aux démonstrations.

Exécutez la cellule suivante une seule fois si le modèle n’est pas encore installé. Dans certains environnements, le téléchargement peut être bloqué par les règles réseau.


```python
import sys

# Installer le modèle anglais de petite taille pour spaCy
!{sys.executable} -m spacy download en_core_web_sm
```

```python
import spacy

# Charger le modèle anglais
nlp = spacy.load("en_core_web_sm")

print("Version de spaCy :", spacy.__version__)
print("Type de l'objet nlp :", type(nlp))
```

### 3.2 Traiter un texte avec spaCy

L’objet `nlp` représente le pipeline. Lorsqu’on lui passe un texte, spaCy crée un objet `Doc` contenant les tokens et les annotations linguistiques produites par le modèle.


```python
text = "Apple is opening a new office in London."

# Appliquer le pipeline spaCy au texte
doc = nlp(text)

# Afficher chaque token et quelques annotations
for token in doc:
    print(
        f"Texte : {token.text:12} | "
        f"Lemme : {token.lemma_:12} | "
        f"Catégorie grammaticale : {token.pos_:6} | "
        f"Entité : {token.ent_type_}"
    )
```

### 3.3 Reconnaissance d'entités nommées (NER)

La reconnaissance d’entités nommées identifie, lorsque le modèle le permet, des éléments comme les personnes, les organisations, les lieux ou les dates. Les prédictions ne sont pas infaillibles : elles doivent être vérifiées.


```python
text = "Microsoft opened an office in Paris in 2024."
doc = nlp(text)

for ent in doc.ents:
    print(f"{ent.text:20} -> {ent.label_}")
```

## 4. Expressions régulières avec `re`

Le module standard `re` permet de rechercher des motifs dans du texte. Il ne réalise pas à lui seul une analyse linguistique complète, mais il est utile pour extraire des éléments suivant une forme précise, comme des adresses e-mail ou des nombres.


```python
import re

text = "Contact us at nlp-course@example.com or support@example.org."

# Trouver les chaînes qui ressemblent à des adresses e-mail
emails = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)

print("Adresses trouvées :", emails)
```
